```python
import pandas as pd
import warnings
from mlxtend.frequent_patterns import apriori, association_rules
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
df = pd.read_csv("basket_supermercado_1000.csv")

# Removendo a coluna transacao
df = df.drop(columns=["transacao"])
df
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.3,
    use_colnames=True
)

print("Sub-conjuntos:")
print(frequent_itemsets.sort_values(by="support", ascending=False))
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)

print("Regras de associação")
print(rules)
```

```python
rules_filtered = rules[
    (rules["lift"] > 1) &
    (rules["confidence"] > 0.7)
].sort_values(by="lift", ascending=False)

print("Regras de associação filtradas")
print(rules_filtered[["antecedents", "consequents", "support", "confidence", "lift"]])
```
