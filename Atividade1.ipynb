```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Coleta de Dados
df = pd.read_csv("marketing (1).csv")
df
```

```python
#df = pd.read_csv("marketing.csv")

#Variaveis
x = df[['anuncios']]
y = df[['cliques']]

#Treinando o modelo Regressão Linear
modelo = LinearRegression()
modelo.fit(x,y)

#Coeficientes
print(f'Inclinação: {modelo.coef_[0]}')
print(f'Inteceptio: {modelo.intercept_}')

#Grafico
y_pred = modelo.predict(x)
plt.scatter(df['anuncios'], df['cliques'], label= "Dados Reais")
plt.plot(df['anuncios'],y_pred,label = "Regressao Linear", color="red")
plt.xlabel("Anuncios")
plt.ylabel("Cliques")
plt.legend()
plt.show()

#Previsão
# cliques = [[10]]
# anuncios_previsto = modelo.predict(cliques)
# print(f'Para tantos cliques de um anuncio {cliques} cliques, a previsão de anuncios é {anuncios_previsto}.')
```
